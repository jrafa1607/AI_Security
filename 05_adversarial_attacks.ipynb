# Adversarial Attacks and Robustness Testing

This notebook demonstrates adversarial perturbation with a tiny linear classifier. A small change to a feature vector can cross a decision boundary. All values are synthetic and the example is for controlled robustness education.

**Learning goals**
- Calculate a linear classifier's score and decision boundary.
- Construct a small perturbation that flips a prediction.
- Compare clean accuracy with perturbed accuracy on a toy set.

```python
# Binary classifier: score = w dot x + b; predict 1 when score >= 0.
weights = (1.0, 1.0)
bias = -1.0

def score(x):
    return sum(w * value for w, value in zip(weights, x)) + bias

def predict(x):
    return int(score(x) >= 0)

sample = (0.55, 0.50)
print("Input:", sample, "score:", score(sample), "prediction:", predict(sample))
```

## 1. A small boundary-crossing perturbation

The original point is just below the decision boundary. A small increase in both coordinates changes the predicted class. This illustrates sensitivity; it does not imply every real model can be attacked this way.

```python
delta = (0.03, 0.03)
perturbed = tuple(x + d for x, d in zip(sample, delta))
print("Perturbed input:", perturbed)
print("Perturbed score:", score(perturbed))
print("Perturbed prediction:", predict(perturbed))
```

## 2. A tiny robustness check

Evaluate a few labeled points before and after applying the same bounded perturbation. Accuracy on a tiny synthetic set is only a demonstration, not a meaningful benchmark.

```python
dataset = [
    ((0.1, 0.2), 0), ((0.3, 0.4), 0), ((0.55, 0.50), 0),
    ((0.8, 0.7), 1), ((0.9, 0.9), 1),
]

def accuracy(rows, transform=lambda x: x):
    return sum(predict(transform(x)) == y for x, y in rows) / len(rows)

def bounded_shift(x):
    epsilon = 0.03
    return tuple(min(1.0, value + epsilon) for value in x)

print(f"Clean accuracy: {accuracy(dataset):.0%}")
print(f"Shifted accuracy: {accuracy(dataset, bounded_shift):.0%}")
```

## 3. Defensive testing ideas

Evaluate diverse, realistic perturbations and report results by class and subgroup. Consider adversarial training, input constraints, calibrated uncertainty, and human review where errors carry high cost. A defense should be tested against adaptive attacks.

## Takeaways

- Adversarial examples expose model sensitivity near decision boundaries.
- Robustness is a measured property under a defined threat model, not a general guarantee.
- Test realistic transformations and report both accuracy and failure modes.
- This notebook's linear classifier is intentionally simple; complex models require more careful evaluation.
