# AI Hallucination: Confident Answers vs. Evidence

This notebook demonstrates how a language-model-like system can produce a plausible but unsupported answer, and how grounding and abstention reduce the risk.

**Learning goals**
- Distinguish fluent text from verified facts.
- Compare an unconstrained answer with one grounded in supplied evidence.
- Practice a simple evidence-checking workflow.

All examples are fictional and deterministic. This is a teaching simulation, not a real LLM.

```python
# A fictional question and a deliberately incomplete evidence packet
question = "In what year did the fictional Northstar Archive open?"
evidence = {
    "location": "Lumen City",
    "founder": "Mira Vale",
    "opening_year": None,  # the source does not state the year
}
print(question)
print("Evidence:", evidence)
```

## 1. A plausible unsupported answer

The function below stands in for a system that fills gaps with a plausible-sounding completion. Its answer is intentionally fabricated; the important point is that confidence and detail do not make it evidence.

```python
def ungrounded_answer(_question):
    return "The Northstar Archive opened in 2018, following a three-year restoration."

answer = ungrounded_answer(question)
print(answer)
```

## 2. Ground the response and abstain when information is missing

A safer response uses only the provided evidence. If the needed fact is absent, it says so instead of guessing.

```python
def grounded_answer(evidence):
    year = evidence.get("opening_year")
    if year is None:
        return "The supplied evidence does not specify the opening year."
    return f"The Northstar Archive opened in {year}."

print(grounded_answer(evidence))
```

## 3. Check claims against the source

Break a response into claims, then label each one based on whether the packet supports it. This small example is not a general fact-checker; real systems need reliable sources, provenance, and human review for important claims.

```python
claims = [
    ("The archive is in Lumen City.", evidence["location"] == "Lumen City"),
    ("Mira Vale founded the archive.", evidence["founder"] == "Mira Vale"),
    ("It opened in 2018.", evidence["opening_year"] == 2018),
]
for claim, supported in claims:
    print(f"{'SUPPORTED' if supported else 'NOT ESTABLISHED'}: {claim}")
```

## Takeaways

- A plausible answer can be false or unsupported.
- Supplying evidence helps only when the system actually follows it.
- Explicitly stating uncertainty or abstaining is often better than inventing details.
- In production, measure factual accuracy and citation support, not just fluency.
